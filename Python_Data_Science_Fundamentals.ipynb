```python
import matplotlib.pyplot as plt

marks = [85, 90, 78]

plt.plot(marks)
plt.title("Student Marks")
plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.show()
```

```python
students = ["Amit", "Priya", "Rahul"]
marks = [85, 90, 78]

plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import pandas as pd

students = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha"],
    "Math": [85, 90, 75, 88],
    "Science": [80, 95, 70, 92]
}

df = pd.DataFrame(students)

print(df)

print("\nAverage Marks")
print(df[["Math", "Science"]].mean())

print("\nHighest Math Marks")
print(df["Math"].max())
```
